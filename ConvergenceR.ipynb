### Tests for DE-MC

## Gelman-Rubin method

Blocked Gibbs uses two methods of checking for convergence, Heidelberger and Welch's diagnostic and Geweke's diagnostic. I plan on adding an additional approach later rather than relying on a single approach after a bit more research into the other approaches (like ESS, EMD, autocorrelation).

```python
%%R
# Set number of iterations to uset the number of iterations to keep from each chain
n <- 2500

# Get list of all chain files
files <- list.files("/content/demc_convergence", pattern = "\\.txt$", full.names = TRUE)
boa.init()

# Read each file into its own chain
for (i in seq_along(files)) {
  chain_data <- as.matrix(read.table(files[i], sep = ",", header = FALSE))
  # Only keep the first n iterations of each chain (easier testing of different N values)
  chain_data <- head(chain_data, n)

  chain_name <- paste0("chain_", i)

  # Save chain to variable (use chain name based on chain number)
  assign(chain_name, chain_data)

  # Import the chain into BOA using variable name
  boa.chain.import(chain_name, type = "S")
}

# Output Gelman and Rubin's diagnostic
boa.print.gandr()
```
